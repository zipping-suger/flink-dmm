```python
import csv
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = []
y = []
lv = []
itr = []
t = []
with open("res.csv") as f:
    reader = csv.reader(f)
    for row in reader:
        x.append(float(row[0]))
        y.append(float(row[1]))
        lv.append(float(row[2]))
        itr.append(float(row[3]))
        t.append(float(row[4]))

x = np.array(x)
y = np.array(y)
lv = np.array(lv)
itr = np.array(itr)
t = np.array(t)
```

```python
plt.hist(lv[itr<10], bins=100)
plt.show()

```

```python
plt.hist(lv[itr>50], bins=1000, alpha=0.5, density=True)
plt.hist(lv[itr<=50], bins=1000, alpha=0.5, density=True)
plt.show()

plt.hist(itr, bins=100, alpha=0.5)
plt.show()

(0.07558852434158325, 0.4858311116695404)
```

```python
rad = (0.07558852434158325, 0.4858311116695404)
a = np.vstack([x, y])
cond = np.linalg.norm(a - np.array(rad)[:, None], axis=0) <= 0.5
plt.hist(lv[cond], bins=1000, alpha=0.5, density=True)
plt.show()

plt.scatter(x[cond], y[cond], c=lv[cond], s=0.1, cmap="jet")
circle = plt.Circle((0.07558852434158325, 0.4858311116695404), 0.5, color='r', fill=False)
plt.gca().add_patch(circle)

plt.show()
```
